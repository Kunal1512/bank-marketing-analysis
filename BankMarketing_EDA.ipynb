```python
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_csv('/content/drive/MyDrive/analytics_projects/bank.csv')
print(df.shape) # row x column
```

```python
print(df.head()) # sample rows
```

```python
print(df.dtypes) # data types
```

```python
print(df.isnull().sum()) # no of null records in each column
```

```python
print(df["deposit"].value_counts()) # details of particular column
```

```python
print(df["deposit"].value_counts(normalize=True).round(3)) # details of particular column
```

```python
print(df.groupby("job")["deposit"].value_counts(normalize=True).round(2)) # details of particular column grouped by another column
```

```python
df["age_group"]= pd.cut(df["age"],bins = [0,30,40,50,100],labels = ["<30","30-40","40-50","50+"])
print(df.groupby("age_group")["deposit"].value_counts(normalize=True).round(3))
print()
print(df.groupby("deposit")["balance"].mean(),round(0))
```

```python
sub_rate = (df["deposit"] == "yes").mean()
print(f"Overall Subscription rate: {sub_rate:0.2%}")
```
